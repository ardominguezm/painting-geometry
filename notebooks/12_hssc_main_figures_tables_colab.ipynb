# HSSC main figures and tables

Rebuilds the frozen full-corpus display items for **The geometry of artistic style across spatial scales**.

Outputs:
- Figure 2: complementarity under unseen-artist generalisation
- Figure 3: multiscale discrimination
- Figure 4: style organisation across scales
- Table 1: dataset/protocol summary
- Table 2: confirmatory contrasts

The builder reads the final Phase VII CSVs and the frozen Phase V style-geometry summary directly from Google Drive and exports PNG/PDF/SVG plus CSV/LaTeX tables.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import json, os, pathlib, shutil, urllib.request

REPO = pathlib.Path('/content/painting-geometry')
BRANCH = 'multiscale-corpus-analysis'
REPO_SLUG = 'ardominguezm/painting-geometry'
if REPO.exists():
    shutil.rmtree(REPO)
(REPO / 'scripts').mkdir(parents=True, exist_ok=True)

# Resolve the current branch commit for a traceable, reproducible download.
api_url = f'https://api.github.com/repos/{REPO_SLUG}/branches/{BRANCH}'
request = urllib.request.Request(api_url, headers={'User-Agent': 'painting-geometry-colab'})
try:
    with urllib.request.urlopen(request, timeout=60) as response:
        commit_sha = json.load(response)['commit']['sha']
    script_url = (
        f'https://raw.githubusercontent.com/{REPO_SLUG}/{commit_sha}/'
        'scripts/build_hssc_main_figures_tables.py'
    )
    script_path = REPO / 'scripts' / 'build_hssc_main_figures_tables.py'
    urllib.request.urlretrieve(script_url, script_path)
except Exception as exc:
    raise RuntimeError(
        'Could not download the Figure/Table builder from GitHub. '
        'Restart the Colab runtime and rerun this cell. '
        f'Original error: {exc}'
    ) from exc

os.chdir(REPO)
print('Branch commit:', commit_sha)
print('Builder downloaded to:', script_path)
```

```python
!python -u scripts/build_hssc_main_figures_tables.py \
  --drive-root /content/drive/MyDrive \
  --output-root /content/drive/MyDrive/painting_geometry_phase7_full/paper_hssc
```

```python
from pathlib import Path
from IPython.display import display, Image, Markdown
import pandas as pd
OUT = Path('/content/drive/MyDrive/painting_geometry_phase7_full/paper_hssc')
for p in sorted((OUT/'figures').glob('*.png')):
    display(Markdown(f'### {p.stem}'))
    display(Image(filename=str(p), width=1100))
for p in sorted((OUT/'tables').glob('*.csv')):
    display(Markdown(f'### {p.stem}'))
    display(pd.read_csv(p))
print('\nPackage:', OUT/'HSSC_main_figures_tables.zip')
```
